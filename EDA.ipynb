```python
df["FlightDate"] = pd.to_datetime(df["FlightDate"])

df[["FlightDate", "Day_Of_Week"]].drop_duplicates().head(20)
```

```python
df["pandas_day_of_week"] = df["FlightDate"].dt.dayofweek + 1
(df["Day_Of_Week"] == df["pandas_day_of_week"]).mean()
```

Value '1' in Day_Of_Week is Monday not Sunday
